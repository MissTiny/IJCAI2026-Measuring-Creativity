```python
def flatten(obj):
    for x in obj:
        if isinstance(x, (list, np.ndarray)):  # handle both
            yield from flatten(x)
        else:
            yield x
```

```python
prior_knowledge = list(flatten(prior.Pairs.tolist()))
```

```python
newness_binarys=[]
current_year = 1960
for i in range(Annotated_cleaned.shape[0]):
    if i%1000==0:
        print(f"{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}: {i}")
    row = Annotated_cleaned.iloc[i]
    newness_binary=[]
    if row.Year <current_year: # If it is prior, skip
        newness_binary=[0 for move in range(row.PlayCount)]
    if row.Year > current_year: # Update prior when year changed
        new_knowledge = list(flatten(Annotated_cleaned[Annotated_cleaned["Year"]==current_year].Pairs.tolist()))
        prior_knowledge.extend(new_knowledge)
        current_year = row.Year #Update Year
        
    if row.Year == current_year: # If it is posterior
        newness_binary = [0 if j in prior_knowledge else 1 for j in row.Pairs ]
    newness_binarys.append(newness_binary)
print(f"{datetime.now().strftime('%Y-%m-%d %H:%M:%S')}: End")
```

```python
Annotated_cleaned["binary_newness"] = newness_binarys
```
